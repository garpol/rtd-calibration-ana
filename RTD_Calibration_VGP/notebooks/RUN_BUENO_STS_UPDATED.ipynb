# 🔬 RUN_BUENO-STS: Análisis de Runs Individuales (Parametrizable)

**Notebook flexible para analizar runs individuales de cualquier tipo de sensor.**

## 📋 Uso rápido

1. **Ejecuta celda 1**: Importa módulos
2. **Modifica celda 2**: Ajusta `FILENAME_PATTERN`, `CALIBSET_PATTERN`, `TEST_FILENAME`
3. **Ejecuta celdas 3-6**: Analiza el run

## 🎯 Ejemplos de configuración

| Tipo de sensor | FILENAME_PATTERN | CALIBSET_PATTERN |
|---------------|------------------|------------------|
| **Frame/STS** | `'STS'` | `'FRAME_SET'` |
| **LAr normal** | `'rtd_set'` | `''` (vacío) |
| **Pre-calibraciones** | `'_pre'` | `''` |
| **Set específico** | `'STS'` | `'FRAME_SET11'` |
