```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import datetime

def find_plateaus(values, times, tolerance=0.01, min_plateau_length=500):
    """
    Encuentra plateaus en los datos.
    - values: array de valores.
    - times: array de tiempos (segundos desde el epoch).
    - tolerance: diferencia máxima para considerar un plateau.
    - min_plateau_length: duración mínima del plateau en segundos.
    """
    plateaus = []
    start_idx = 0

    for i in range(1, len(values)):
        if abs(values[i] - values[start_idx]) > tolerance:
            duration = times[i - 1] - times[start_idx]
            if duration >= min_plateau_length:
                segment = values[start_idx:i]
                plateaus.append({
                    "t0": times[start_idx],
                    "tfin": times[i - 1],
                    "mean": np.mean(segment),
                    "std": np.std(segment),
                    "data": segment
                })
            start_idx = i  # Reiniciar seguimiento
    
    return plateaus

def calculate_plateaus(input_filename, sensor_temp=0, sensor_wav=(0, 0), params=None):
    """
    Calcula los plateaus para 'press', 'temp' y 'wav' de manera independiente.
    input_filename: ruta del archivo ROOT con los datos.
    sensor_temp: índice del sensor de temperatura a considerar.
    sensor_wav: tupla (x, y) del sensor de wav a considerar.
    params: diccionario con parámetros específicos para cada rama {"press": {...}, "temp": {...}, "wav": {...}}
    """
    if params is None:
        params = {
            "press": {"tolerance": 0.01, "min_plateau_length": 500},
            "temp": {"tolerance": 0.01, "min_plateau_length": 500},
            "wav": {"tolerance": 0.01, "min_plateau_length": 500}
        }
    
    with uproot.open(input_filename) as file:
        tree = file["resampled_data"]
        data = tree.arrays(library="np")
    
    times = data["t"]
    timestamps = np.array([datetime.datetime.utcfromtimestamp(float(t)) for t in times])
    
    press_plateaus = find_plateaus(data['press'], times, **params["press"])
    temp_plateaus = find_plateaus([row[sensor_temp] for row in data['temp']], times, **params["temp"])

    # Filtrar los valores de 'wav' según los umbrales
    wav_ = [row[sensor_wav[0]][sensor_wav[1]] for row in data['wav']]
    umbral_min = 1.5574e-6
    umbral_max = 1.5578e-6
    mask_wav = (np.array(wav_) > umbral_min) & (np.array(wav_) < umbral_max)
    
    # Filtrar los datos y los tiempos
    timestamps_filtrados = np.array(timestamps)[mask_wav]
    wav_filtrado = np.array(wav_)[mask_wav]
    
    # Ahora calculamos los plateaus para los datos filtrados de 'wav'
    wav_plateaus = find_plateaus(wav_filtrado*1e6, timestamps_filtrados, **params["wav"])
    
    # Graficar los datos con los plateaus por separado
    for branch, series, plateaus, color in zip(["press", "temp", "wav"],
                                                [data['press'], [row[sensor_temp] for row in data['temp']], wav_filtrado*1e6],
                                                [press_plateaus, temp_plateaus, wav_plateaus],
                                                ["blue", "green", "orange"]):
        plt.figure(figsize=(10, 5))
        if branch == "wav":
            # En el caso de 'wav', utilizamos los tiempos y valores filtrados
            plt.plot(timestamps_filtrados, series, marker='o', linestyle='-', label=branch, color=color)
        else:
            # Para 'press' y 'temp', se grafican con los tiempos originales
            plt.plot(timestamps, series, marker='o', linestyle='-', label=branch, color=color)
        
        for plateau in plateaus:
            plt.axvline(x=datetime.datetime.utcfromtimestamp(plateau["t0"]), color='red', linestyle='--', alpha=0.5)
            plt.axvline(x=datetime.datetime.utcfromtimestamp(plateau["tfin"]), color='blue', linestyle='--', alpha=0.5)
        
        plt.xlabel("Time (UTC)")
        plt.ylabel(branch)
        plt.title(f"{branch} vs Time with Plateaus")
        plt.xticks(rotation=45)
        plt.legend()
        plt.grid()
        plt.show()
    
    # Imprimir plateaus de cada rama
    for branch, plateaus in {"press": press_plateaus, "temp": temp_plateaus, "wav": wav_plateaus}.items():
        print(f"\nPlateaus de {branch} detectados:")
        for i, plateau in enumerate(plateaus):
            print(f"Plateau {i}: t0={datetime.datetime.utcfromtimestamp(plateau['t0'])}, "
                  f"tfin={datetime.datetime.utcfromtimestamp(plateau['tfin'])}, "
                  f"mean={plateau['mean']:.6e}, std={plateau['std']:.6e}, "
                  f"data_size={len(plateau['data'])}")
    
    return {"press": press_plateaus, "temp": temp_plateaus, "wav": wav_plateaus}

params = {
    "press": {"tolerance": 0.25, "min_plateau_length": 1000},
    "temp": {"tolerance": 0.5, "min_plateau_length": 800},
    "wav": {"tolerance": 1e-2, "min_plateau_length": 1000}
}

# Ejecutar el cálculo de plateaus
plateaus_result = calculate_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_temp=7, sensor_wav=(0, 0), params=params)

#para wav lo hacemos a parte porque la tolerancia es de 1e-5 y da fallos si no cambiamos la estructura de tiempos 
```

```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import datetime

def find_plateaus(values, times, tolerance, min_plateau_length):
    plateaus = []
    start_idx = 0

    for i in range(1, len(values)):
        if abs(values[i] - values[start_idx]) > tolerance:
            duration = times[i - 1] - times[start_idx]

            # Convertir 'duration' de timedelta a segundos
            duration_seconds = duration.total_seconds()

            # Si la duración es suficientemente larga, guardar el plateau
            if duration_seconds >= min_plateau_length:
                segment = values[start_idx:i]

                plateaus.append({
                    "t0": times[start_idx],  # Usar directamente times[start_idx]
                    "tfin": times[i - 1],    # Usar directamente times[i - 1]
                    "duration": duration_seconds,
                    "values": segment
                })

            start_idx = i

    return plateaus

def calculate_wav_plateaus(input_filename, sensor_wav=(0, 0), params=None):
    """
    Calcula los plateaus para 'wav' de manera independiente.
    input_filename: ruta del archivo ROOT con los datos.
    sensor_wav: tupla (x, y) del sensor de wav a considerar.
    params: diccionario con parámetros específicos para la rama "wav".
    """
    if params is None:
        params = {
            "wav": {"tolerance": 0.2, "min_plateau_length": 500}
        }
    
    with uproot.open(input_filename) as file:
        tree = file["resampled_data"]
        data = tree.arrays(library="np")
    
    times = data["t"]
    timestamps = np.array([datetime.datetime.utcfromtimestamp(float(t)) for t in times])

    # Filtrar los valores de 'wav' según los umbrales
    wav_ = data["wav"][:, sensor_wav[0], sensor_wav[1]]
    umbral_min = 1.5574e-6
    umbral_max = 1.5578e-6
    mask_wav = (np.array(wav_) > umbral_min) & (np.array(wav_) < umbral_max)
    
    # Filtrar los datos y los tiempos
    timestamps_filtrados = np.array(timestamps)[mask_wav]
    wav_filtrado = np.array(wav_)[mask_wav]
    
    # Ahora calculamos los plateaus para los datos filtrados de 'wav'
    wav_plateaus = find_plateaus(wav_filtrado*1e6, timestamps_filtrados, **params["wav"])
    
    # Graficar los datos de 'wav' con los plateaus
    plt.figure(figsize=(10, 5))
    plt.plot(timestamps_filtrados, wav_filtrado*1e6, marker='o', linestyle='-', label="wav", color="orange")
    
    for plateau in wav_plateaus:
        plt.axvline(x=plateau["t0"], color='red', linestyle='--', alpha=0.5)
        plt.axvline(x=plateau["tfin"], color='blue', linestyle='--', alpha=0.5)
    
    plt.xlabel("Time (UTC)")
    plt.ylabel("wav x 10^6")
    plt.title("wav vs Time with Plateaus")
    plt.xticks(rotation=45)
    plt.legend()
    plt.grid()
    plt.show()
    
    # Imprimir los plateaus detectados para 'wav'
    print(f"\nPlateaus de 'wav' detectados:")
    for i, plateau in enumerate(wav_plateaus):
        print(f"Plateau {i}: t0={plateau['t0']}, "
              f"tfin={plateau['tfin']}, "
              f"mean={np.mean(plateau['values']):.6e}, std={np.std(plateau['values']):.6e}, "
              f"data_size={len(plateau['values'])}")
    
    return {"wav": wav_plateaus}


params = {
    "wav": {"tolerance": 0.000015, "min_plateau_length": 1000}
}
print(len(timestamps_filtrados), len(wav_filtrado))  # Verifica que ambas arrays tengan la misma longitud

# Ejecutar el cálculo de plateaus
wav_plateaus_result_00 = calculate_wav_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_wav=(0, 0), params=params)

params = {
    "wav": {"tolerance": 0.000008, "min_plateau_length": 1000}
}


wav_plateaus_result_10 = calculate_wav_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_wav=(1, 0), params=params)
```

```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import datetime

def find_plateaus(values, times, tolerance=0.01, min_plateau_length=500):
    """
    Encuentra plateaus en los datos.
    - values: array de valores.
    - times: array de tiempos (segundos desde el epoch).
    - tolerance: diferencia máxima para considerar un plateau.
    - min_plateau_length: duración mínima del plateau en segundos.
    """
    plateaus = []
    start_idx = 0

    for i in range(1, len(values)):
        if abs(values[i] - values[start_idx]) > tolerance:
            duration = times[i - 1] - times[start_idx]
            duration_seconds = duration.total_seconds()  # Convertir a segundos
            if duration_seconds >= min_plateau_length:
                segment = values[start_idx:i]
                plateaus.append({
                    "t0": times[start_idx],
                    "tfin": times[i - 1],
                    "mean": np.mean(segment),
                    "std": np.std(segment),
                    "data": segment
                })
            start_idx = i  # Reiniciar seguimiento
    
    return plateaus

def calculate_temp_plateaus(input_filename, sensor_temp=0, params=None):
    """
    Calcula los plateaus para 'temp' de manera independiente.
    input_filename: ruta del archivo ROOT con los datos.
    sensor_temp: índice del sensor de temperatura a considerar.
    params: diccionario con parámetros específicos para cada rama {"temp": {...}}
    """
    if params is None:
        params = {
            "temp": {"tolerance": 0.01, "min_plateau_length": 500}
        }
    
    with uproot.open(input_filename) as file:
        tree = file["resampled_data"]
        data = tree.arrays(library="np")
    
    times = data["t"]
    timestamps = np.array([datetime.datetime.utcfromtimestamp(float(t)) for t in times])
    
    # Filtro de tiempo: Seleccionar solo los datos a partir de las 14:00 (hora UTC)
    filter_time = datetime.time(13, 46)  # 14:00 UTC
    timestamps_filtrados = [t for t in timestamps if t.time() >= filter_time]
    temp_filtrado = [row[sensor_temp] for row, t in zip(data['temp'], timestamps) if t in timestamps_filtrados]

    # Calculamos los plateaus de los datos filtrados de temperatura
    temp_plateaus = find_plateaus(temp_filtrado, timestamps_filtrados, **params["temp"])

    # Graficar los datos de temperatura con los plateaus
    plt.figure(figsize=(10, 5))
    plt.plot(timestamps_filtrados, temp_filtrado, marker='o', linestyle='-', label="Data 27-02-2028", color="green")
    
    for plateau in temp_plateaus:
        plt.axvline(x=plateau["t0"], color='red', linestyle='--', alpha=0.5)
        plt.axvline(x=plateau["tfin"], color='blue', linestyle='--', alpha=0.5)
    
    plt.xlabel("Time (UTC)")
    plt.ylabel("Temperature (K)")
    plt.title(f"Temperature Plateaus (Sensor {sensor_temp})")
    plt.xticks(rotation=45)
    plt.legend()
    plt.grid()
    plt.show()
    
    # Imprimir plateaus de temperatura
    print(f"\nPlateaus de temperatura detectados:")
    for i, plateau in enumerate(temp_plateaus):
        print(f"Plateau {i}: t0={plateau['t0']}, "
              f"tfin={plateau['tfin']}, "
              f"mean={plateau['mean']:.6e}, std={plateau['std']:.6e}, "
              f"data_size={len(plateau['data'])}")
    
    return {"temp": temp_plateaus}

params = {
    "temp": {"tolerance": 0.25, "min_plateau_length": 1000}   
}

# Ejecutar el cálculo de plateaus
temp_plateaus_result_7 = calculate_temp_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_temp=7, params=params)

params = {
    "temp": {"tolerance": 0.4, "min_plateau_length": 1000}   
}

temp_plateaus_result_6 = calculate_temp_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_temp=6, params=params)
```

```python
def find_plateaus(values, times, tolerance, min_plateau_length):
    plateaus = []
    start_idx = 0

    for i in range(1, len(values)):
        if abs(values[i] - values[start_idx]) > tolerance:
            duration = times[i - 1] - times[start_idx]

            # Convertir 'duration' de timedelta a segundos
            duration_seconds = duration.total_seconds()

            # Si la duración es suficientemente larga, guardar el plateau
            if duration_seconds >= min_plateau_length:
                segment = values[start_idx:i]

                plateaus.append({
                    "t0": times[start_idx],  # Usar directamente times[start_idx]
                    "tfin": times[i - 1],    # Usar directamente times[i - 1]
                    "duration": duration_seconds,
                    "values": segment
                })

            start_idx = i

    return plateaus


def calculate_wav_plateaus(input_filename, sensor_wav=(0, 0), params=None):
    """
    Calcula los plateaus para 'wav' de manera independiente.
    input_filename: ruta del archivo ROOT con los datos.
    sensor_wav: tupla (x, y) del sensor de wav a considerar.
    params: diccionario con parámetros específicos para la rama "wav".
    """
    if params is None:
        params = {
            "wav": {"tolerance": 0.2, "min_plateau_length": 500}
        }
    
    with uproot.open(input_filename) as file:
        tree = file["resampled_data"]
        data = tree.arrays(library="np")
    
    times = data["t"]
    timestamps = np.array([datetime.datetime.utcfromtimestamp(float(t)) for t in times])

    # Filtrar los valores de 'wav' según los umbrales
    wav_ = data["wav"][:, sensor_wav[0], sensor_wav[1]]
    
    umbral_min = 1.563e-6
    umbral_max = 1.5642e-6
    mask_wav = (np.array(wav_) > umbral_min) & (np.array(wav_) < umbral_max)
    
    # Filtrar los datos y los tiempos
    timestamps_filtrados = np.array(timestamps)[mask_wav]
    wav_filtrado = np.array(wav_)[mask_wav]
    
    # Ahora calculamos los plateaus para los datos filtrados de 'wav'
    wav_plateaus = find_plateaus(wav_filtrado*1e6, timestamps_filtrados, **params["wav"])
    

    
    # Define polarization and sensor labels
    polarization = "P" if sensor_wav[0] == 0 else "S"
    sensor_num = sensor_wav[1] + 1
    
    # Plot 'wav' data with plateaus
    plt.figure(figsize=(10, 5))
    plt.plot(timestamps_filtrados, wav_filtrado*1e6, marker='o', linestyle='-', 
             label=f"Wavelength Shift (Polarization {polarization}, Sensor {sensor_num})", color="orange")
    
    for plateau in wav_plateaus:
        plt.axvline(x=plateau["t0"], color='red', linestyle='--', alpha=0.5)
        plt.axvline(x=plateau["tfin"], color='blue', linestyle='--', alpha=0.5)
        
    plt.xlabel("Time (UTC)", fontsize=14, fontweight='bold')
    plt.ylabel("Wavelength Shift (micrometers)", fontsize=14, fontweight='bold')
    plt.title(f"Wavelength Shift Over Time with Plateaus (Polarization {polarization}, Sensor {sensor_num})", fontsize=16, fontweight='bold')
    plt.xticks(rotation=45, fontsize=12)
    plt.yticks(fontsize=12)
    plt.legend(fontsize=12)
    plt.grid()
    plt.show()
    
    # Imprimir los plateaus detectados para 'wav'
    print(f"\nPlateaus de 'wav' detectados:")
    for i, plateau in enumerate(wav_plateaus):
        print(f"Plateau {i}: t0={plateau['t0']}, "
              f"tfin={plateau['tfin']}, "
              f"mean={np.mean(plateau['values']):.6e}, std={np.std(plateau['values']):.6e}, "
              f"data_size={len(plateau['values'])}")
    
    return {"wav": wav_plateaus}


params = {
    "wav": {"tolerance": 0.000009, "min_plateau_length": 1000}
}
print(len(timestamps_filtrados), len(wav_filtrado))  # Verifica que ambas arrays tengan la misma longitud

# Ejecutar el cálculo de plateaus
wav_plateaus_result_02 = calculate_wav_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_wav=(0, 2), params=params)

params = {
    "wav": {"tolerance": 0.000009, "min_plateau_length": 1000}
}

wav_plateaus_result_12 = calculate_wav_plateaus("/eos/user/j/jcapotor/FBGdata/ROOTFiles/pressure_setup/resampled20250227.root", sensor_wav=(1, 2), params=params)
```
