## 5️⃣ Procesamiento FBG: Filtrado y Media de Polarizaciones

### 5.1 Configuración de 4 Sensores FBG

```python
# ✅ CONFIGURACIÓN: 8 sensores FBG completos (4 + 4)
# 8 sensores × 2 polarizaciones = 16 canales
# Índices 0-based: sensor 0, 1, 2, 3, 4, 5, 6, 7 (Python indexing)
fbg_config = []
for sensor in range(8):
    fbg_config.append({"pol": 0, "sensor": sensor, "label": f"FBG-{sensor+1}-P"})
    fbg_config.append({"pol": 1, "sensor": sensor, "label": f"FBG-{sensor+1}-S"})
print(f"✅ Configurados {len(fbg_config)} canales FBG (8 sensores × 2 polarizaciones)")
print(f"💡 Se verificará si hay datos válidos para todos los sensores en este dataset")
```

### 5.2 Filtrado Wavelength (μ±3σ)

```python
def filter_wavelength_data(wav_data, timestamps, sigma_threshold=3.0, use_robust=True):
    """
    Filtra wavelength usando método μ±3σ o método robusto basado en mediana (MAD)
    
    Parameters:
    -----------
    wav_data : array
        Datos de wavelength a filtrar
    timestamps : array
        Timestamps correspondientes
    sigma_threshold : float
        Número de sigmas (o MADs) para el threshold
    use_robust : bool
        Si True, usa MAD (Median Absolute Deviation) en lugar de std
    """
    mask_positive = wav_data > 0
    wav_positive = wav_data[mask_positive]
    
    # Si no hay datos válidos, retornar estructura vacía
    if len(wav_positive) == 0:
        return {
            "timestamps": np.array([]),
            "wavelength": np.array([]),
            "mean": np.nan,
            "std": np.nan,
            "n_valid": 0
        }
    
    if use_robust:
        # Método robusto: usa MEDIANA y MAD (más resistente a outliers)
        median_wav = np.median(wav_positive)
        mad = np.median(np.abs(wav_positive - median_wav))
        # Convertir MAD a escala comparable con std (factor 1.4826)
        sigma_equiv = 1.4826 * mad
        
        threshold_min = median_wav - sigma_threshold * sigma_equiv
        threshold_max = median_wav + sigma_threshold * sigma_equiv
        
        # Filtrar usando los thresholds robustos
        mask_filtered = (wav_data > 0) & (wav_data >= threshold_min) & (wav_data <= threshold_max)
        
        # Calcular estadísticas finales sobre datos filtrados
        wav_filtered = wav_data[mask_filtered]
        mean_wav = np.mean(wav_filtered)
        std_wav = np.std(wav_filtered)
    else:
        # Método clásico: μ±3σ
        mean_wav = np.mean(wav_positive)
        std_wav = np.std(wav_positive)
        threshold_min = mean_wav - sigma_threshold * std_wav
        threshold_max = mean_wav + sigma_threshold * std_wav
        
        mask_filtered = (wav_data > 0) & (wav_data >= threshold_min) & (wav_data <= threshold_max)
    
    return {
        "timestamps": timestamps[mask_filtered],
        "wavelength": wav_data[mask_filtered],
        "mean": mean_wav,
        "std": std_wav,
        "n_valid": np.sum(mask_filtered)
    }
    
    # DIAGNÓSTICO: Verificar forma y valores de los sensores
    print(f"🔍 Forma de wav_filtered: {wav_filtered.shape}")
    print(f"   Dimensiones: (time={wav_filtered.shape[0]}, pol={wav_filtered.shape[1]}, sensor={wav_filtered.shape[2]})")
    print(f"\n🔍 Diagnóstico de valores por sensor (FBG-1 a FBG-8):")
    for sensor_idx in range(8):
        for pol_idx in range(2):
            pol_name = "P" if pol_idx == 0 else "S"
            sensor_data = wav_filtered[:, pol_idx, sensor_idx]
            n_positive = np.sum(sensor_data > 0)
            if n_positive > 0:
                min_val = np.min(sensor_data[sensor_data > 0])
                max_val = np.max(sensor_data)
                mean_val = np.mean(sensor_data[sensor_data > 0])
            else:
                min_val = max_val = mean_val = 0
            print(f"   FBG-{sensor_idx+1}-{pol_name}: {n_positive:7d} pts > 0, range=[{min_val:.2e}, {max_val:.2e}] nm, μ={mean_val:.2e} nm")
```

### 5.3 Calcular Media de Polarizaciones

```python
# Filtrar datos wavelength para cada sensor y polarización (8 sensores × 2 pol)
# Diagnóstico de shapes y tipos
print(f"wav_raw shape: {wav_raw.shape}")
print(f"times_all shape: {times_all.shape}")
print(f"times_all[0]: {times_all[0]}")

# Usar solo la primera columna de times_all para el tiempo de wavelength (P)
times_wav = times_all[:, 0]  # shape (153125,)
print(f"times_wav shape: {times_wav.shape}, dtype: {times_wav.dtype}, first: {times_wav[0]}")

# Definir t_inicio y t_fin como escalares para la columna usada
t_inicio = times_all[0, 0]
t_fin = times_all[-1, 0]
print(f"t_inicio: {t_inicio}, t_fin: {t_fin}, type: {type(t_inicio)}")

# Si times_wav es float (timestamp), no convertir
times_wav_dt = times_wav

# Crear máscara temporal
mask_time = (times_wav_dt >= t_inicio) & (times_wav_dt <= t_fin)
print(f"mask_time shape: {mask_time.shape}, sum: {np.sum(mask_time)}")

fbg_filtered_data = {}
for config in fbg_config:
    pol = config['pol']
    sensor = config['sensor']
    label = config['label']
    wav_sensor_pol = wav_raw[:, pol, sensor][mask_time]
    timestamps_sensor = times_wav_dt[mask_time]
    filtered = filter_wavelength_data(wav_sensor_pol, timestamps_sensor, sigma_threshold=3.0, use_robust=True)
    fbg_filtered_data[label] = filtered
print(f"✅ Datos filtrados para {len(fbg_filtered_data)} canales (8 sensores × 2 pol)")
```

```python
# Calcular media P+S para cada sensor (8 sensores: 1 a 8)
fbg_mean_data = {}

for sensor_num in range(1, num_sensores + 1):  # Sensores 1 a num_sensores
    label_p = f"FBG-{sensor_num}-P"
    label_s = f"FBG-{sensor_num}-S"
    label_mean = f"FBG-{sensor_num}-Mean"
    d_p = fbg_filtered_data[label_p]
    d_s = fbg_filtered_data[label_s]
    times_p = d_p["timestamps"]
    times_s = d_s["timestamps"]
    wav_p = d_p["wavelength"]
    wav_s = d_s["wavelength"]
    # Verificar que hay datos válidos
    if len(times_p) == 0 or len(times_s) == 0:
        print(f"⚠️ FBG-{sensor_num} no tiene datos suficientes, omitiendo...")
        continue
    # Interpolar S a tiempos de P (usar directamente los float timestamps)
    times_p_sec = np.array(times_p)
    times_s_sec = np.array(times_s)
    wav_s_interp = np.interp(times_p_sec, times_s_sec, wav_s)
    # Media
    wav_mean = (wav_p + wav_s_interp) / 2.0
    fbg_mean_data[label_mean] = {
        "timestamps": times_p,
        "wavelength": wav_mean,
        "mean": np.mean(wav_mean),
        "std": np.std(wav_mean),
        "n_valid": len(wav_mean),
        "sensor_num": sensor_num
    }

print("✅ Medias calculadas (P+S)/2 para todos los sensores")
print(f"\n{'Sensor':<15} {'Mean (nm)':<15} {'Std (nm)':<15} {'N puntos'}")
print("="*65)
for label, d in fbg_mean_data.items():
    print(f"{label:<15} {d['mean']:<15.6e} {d['std']:<15.6e} {d['n_valid']}")
```

```python
# ══════════════════════════════════════════════════════════════════════════════
# 🔬 COMPARACIÓN DE PICOS: Canal 0 vs Canal 1 — Amplitudes normalizadas + FWHM
# ══════════════════════════════════════════════════════════════════════════════
from scipy.signal import find_peaks
from scipy.interpolate import interp1d

spectrum_wav = data['spectrum_wav']   # (n_esp, n_pol, n_chan, n_pts)
n_espectros, n_pol, n_chan, n_pts = spectrum_wav.shape

# Usar media de polarizaciones para cada canal
def mean_pols(chan, esp=0):
    """Devuelve la media (P+S)/2 del espectro dado."""
    return np.mean(spectrum_wav[esp, :, chan, :], axis=0)

# ── Extraer señales ──────────────────────────────────────────────────────────
# Usar espectro índice 0; ajusta idx_esp si quieres otro
idx_esp = 0
sig0 = mean_pols(0, idx_esp)   # Canal 0 — DW-3 Ormocer-T
sig1 = mean_pols(1, idx_esp)   # Canal 1 — POLY-1

# Eje x: índice de punto (sin eje λ calibrado en estos datos)
x = np.arange(n_pts)

# ── Encontrar el primer pico en cada señal ────────────────────────────────────
def find_first_peak(sig, prominence=0.05):
    """Devuelve índice del primer pico con prominencia relativa al rango."""
    sig_norm = (sig - sig.min()) / (sig.max() - sig.min())
    peaks, props = find_peaks(sig_norm, prominence=prominence, distance=50)
    if len(peaks) == 0:
        # Fallback: máximo global
        return np.argmax(sig_norm), sig_norm
    return peaks[0], sig_norm

pk0, sig0_norm = find_first_peak(sig0)
pk1, sig1_norm = find_first_peak(sig1)

# ── FWHM por interpolación lineal ─────────────────────────────────────────────
def calc_fwhm(x, sig_norm, pk_idx):
    """
    Calcula FWHM interpolando cruce de 0.5 a ambos lados del pico.
    Devuelve (fwhm_pts, x_left, x_right).
    """
    half = 0.5
    # Lado izquierdo: último punto < 0.5 antes del pico
    left_region = sig_norm[:pk_idx]
    idxs_left = np.where(left_region < half)[0]
    if len(idxs_left) == 0:
        x_left = float(x[0])
    else:
        i_l = idxs_left[-1]
        # Interpolar entre i_l e i_l+1
        f = interp1d([sig_norm[i_l], sig_norm[i_l+1]], [x[i_l], x[i_l+1]])
        x_left = float(f(half))

    # Lado derecho: primer punto < 0.5 después del pico
    right_region = sig_norm[pk_idx:]
    idxs_right = np.where(right_region < half)[0]
    if len(idxs_right) == 0:
        x_right = float(x[-1])
    else:
        i_r = pk_idx + idxs_right[0]
        # Interpolar entre i_r-1 e i_r
        f = interp1d([sig_norm[i_r-1], sig_norm[i_r]], [x[i_r-1], x[i_r]])
        x_right = float(f(half))

    fwhm = x_right - x_left
    return fwhm, x_left, x_right

fwhm0, xl0, xr0 = calc_fwhm(x, sig0_norm, pk0)
fwhm1, xl1, xr1 = calc_fwhm(x, sig1_norm, pk1)

# ── Plot centrado en el pico (ventana ±200 pts) ───────────────────────────────
half_win = 200
i0_s, i0_e = max(0, pk0 - half_win), min(n_pts, pk0 + half_win)
i1_s, i1_e = max(0, pk1 - half_win), min(n_pts, pk1 + half_win)

# Coordenadas relativas al pico para superponer
x0_rel = x[i0_s:i0_e] - pk0
x1_rel = x[i1_s:i1_e] - pk1

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x0_rel, sig0_norm[i0_s:i0_e], color='royalblue',  linewidth=2.0,
        label=f'Canal 0 — DW-3 Ormocer-T  (FWHM = {fwhm0:.1f} pts)')
ax.plot(x1_rel, sig1_norm[i1_s:i1_e], color='tomato', linewidth=2.0,
        label=f'Canal 1 — POLY-1 Poliamida  (FWHM = {fwhm1:.1f} pts)')

# Marcar FWHM — Canal 0
ax.hlines(0.5, xl0 - pk0, xr0 - pk0,
          colors='royalblue', linestyles='--', linewidth=1.4, alpha=0.7)
ax.axvline(xl0 - pk0, color='royalblue', linestyle=':', linewidth=1.0, alpha=0.5)
ax.axvline(xr0 - pk0, color='royalblue', linestyle=':', linewidth=1.0, alpha=0.5)

# Marcar FWHM — Canal 1
ax.hlines(0.5, xl1 - pk1, xr1 - pk1,
          colors='tomato', linestyles='--', linewidth=1.4, alpha=0.7)
ax.axvline(xl1 - pk1, color='tomato', linestyle=':', linewidth=1.0, alpha=0.5)
ax.axvline(xr1 - pk1, color='tomato', linestyle=':', linewidth=1.0, alpha=0.5)

ax.axhline(0.5, color='gray', linestyle=':', linewidth=0.8, alpha=0.5, label='Nivel 0.5 (FWHM)')
ax.set_xlabel('Índice relativo al pico', fontsize=12)
ax.set_ylabel('Amplitud normalizada', fontsize=12)
ax.set_title(
    f'Comparación primer pico: Canal 0 vs Canal 1\n'
    f'Espectro #{idx_esp}  —  Amplitudes normalizadas (media Pol P+S)',
    fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-half_win, half_win)
ax.set_ylim(-0.05, 1.15)
plt.tight_layout()
plt.show()

# ── Resumen numérico ──────────────────────────────────────────────────────────
print(f"\n{'─'*60}")
print(f"  📊 RESUMEN FWHM — Primer pico (espectro #{idx_esp})")
print(f"{'─'*60}")
print(f"  {'Canal':<8}  {'Fibra':<22}  {'Pico (idx)':>12}  {'FWHM (pts)':>12}")
print(f"  {'-'*58}")
print(f"  {'0':<8}  {'DW-3 Ormocer-T':<22}  {pk0:>12d}  {fwhm0:>12.1f}")
print(f"  {'1':<8}  {'POLY-1 Poliamida':<22}  {pk1:>12d}  {fwhm1:>12.1f}")
print(f"{'─'*60}")
print(f"  Ratio FWHM (Canal1 / Canal0): {fwhm1/fwhm0:.3f}")
print(f"  💡 FWHM mayor → pico más ancho → FBG más corto o menor reflectividad")
print(f"{'─'*60}")
```

```python
# ══════════════════════════════════════════════════════════════════════════════
# 🔬 DIAGNÓSTICO: Por qué POLY-1 tiene FWHM 2.25x mayor que DW-3
#    + Fit Gaussiano propio para centroide más preciso que el interrogador
# ══════════════════════════════════════════════════════════════════════════════
from scipy.optimize import curve_fit

# ── Función Gaussiana ─────────────────────────────────────────────────────────
def gaussian(x, A, mu, sigma, C):
    return A * np.exp(-0.5 * ((x - mu) / sigma)**2) + C

def fit_gaussian_peak(sig_norm, pk_idx, half_win=150):
    """
    Ajusta Gaussiana a la región ±half_win alrededor del pico.
    Devuelve (mu_fit, sigma_fit, fwhm_fit, pcov, x_win, y_win).
    """
    i_s = max(0, pk_idx - half_win)
    i_e = min(len(sig_norm), pk_idx + half_win)
    x_win = np.arange(i_s, i_e, dtype=float)
    y_win = sig_norm[i_s:i_e]

    p0 = [1.0, float(pk_idx), 30.0, 0.0]
    bounds = ([0, pk_idx - half_win, 1, -0.1],
              [1.2, pk_idx + half_win, half_win, 0.2])
    try:
        popt, pcov = curve_fit(gaussian, x_win, y_win, p0=p0, bounds=bounds, maxfev=5000)
        A, mu, sigma, C = popt
        fwhm_fit = 2.355 * abs(sigma)  # FWHM = 2√(2ln2)·σ
        return mu, sigma, fwhm_fit, pcov, x_win, y_win, popt
    except RuntimeError:
        return float(pk_idx), None, None, None, x_win, y_win, None

# ── Fit en los dos canales ────────────────────────────────────────────────────
mu0, sigma0, fwhm_g0, pcov0, xw0, yw0, popt0 = fit_gaussian_peak(sig0_norm, pk0)
mu1, sigma1, fwhm_g1, pcov1, xw1, yw1, popt1 = fit_gaussian_peak(sig1_norm, pk1)

# Incertidumbre del centroide (1σ del fit)
unc0 = np.sqrt(pcov0[1,1]) if pcov0 is not None else np.nan
unc1 = np.sqrt(pcov1[1,1]) if pcov1 is not None else np.nan

# ── Plot comparativo con fits ─────────────────────────────────────────────────
half_win_plot = 200
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax_i, (sig_n, pk, mu, sigma, fwhm_g, popt, xw, yw, label_txt, color) in enumerate([
    (sig0_norm, pk0, mu0, sigma0, fwhm_g0, popt0, xw0, yw0, 'Canal 0 — DW-3 Ormocer-T',  'royalblue'),
    (sig1_norm, pk1, mu1, sigma1, fwhm_g1, popt1, xw1, yw1, 'Canal 1 — POLY-1 Poliamida', 'tomato'),
]):
    ax = axes[ax_i]
    i_s = max(0, pk - half_win_plot)
    i_e = min(len(sig_n), pk + half_win_plot)
    x_rel = np.arange(i_s, i_e) - pk
    ax.plot(x_rel, sig_n[i_s:i_e], color=color, lw=2.0, label='Espectro (media P+S)')

    if popt is not None:
        x_fit = np.linspace(i_s, i_e, 500)
        y_fit = gaussian(x_fit, *popt)
        ax.plot(x_fit - pk, y_fit, 'k--', lw=1.5, label=f'Fit Gaussiano\nFWHM = {fwhm_g:.1f} pts\nμ = {mu - pk:+.2f} pts del máx')
        ax.axvline(mu - pk, color='k', lw=0.8, linestyle=':')

    ax.axhline(0.5, color='gray', lw=0.8, linestyle=':', alpha=0.6)
    ax.set_title(label_txt, fontsize=12)
    ax.set_xlabel('Índice relativo al pico', fontsize=11)
    ax.set_ylabel('Amplitud normalizada', fontsize=11)
    ax.set_xlim(-half_win_plot, half_win_plot)
    ax.set_ylim(-0.05, 1.15)
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('Fit Gaussiano a los picos — Canal 0 vs Canal 1', fontsize=13)
plt.tight_layout()
plt.show()

# ── Resumen y diagnóstico ─────────────────────────────────────────────────────
N_GROUP = 1.468
LAM_B = 1550.0  # nm nominal (sin calibración λ exacta)
pts_per_nm = (n_pts / 80.0)  # ~39200 pts / 80 nm de rango típico del interrogador

fwhm0_nm = fwhm_g0 / pts_per_nm if fwhm_g0 else None
fwhm1_nm = fwhm_g1 / pts_per_nm if fwhm_g1 else None

def L_est(fwhm_nm):
    """Longitud estimada del FBG (mm) a partir de FWHM (nm)."""
    return (LAM_B**2 / (2 * N_GROUP * fwhm_nm)) / 1e6

print(f"\n{'═'*65}")
print(f"  🔬 DIAGNÓSTICO — Por qué POLY-1 tiene pico más ancho")
print(f"{'═'*65}")
print(f"\n  Fit Gaussiano (σ Gaussiana ↔ FWHM = 2.355·σ):")
print(f"  {'Canal':<8}  {'Fibra':<22}  {'FWHM (pts)':>12}  {'FWHM (nm)~':>12}  {'L_FBG~(mm)':>12}  {'Unc. μ (pts)':>14}")
print(f"  {'-'*82}")
for label_c, fwhm_g, fwhm_nm_val, unc in [
    ('0 DW-3',   fwhm_g0, fwhm0_nm, unc0),
    ('1 POLY-1', fwhm_g1, fwhm1_nm, unc1),
]:
    L = f'{L_est(fwhm_nm_val):.2f}' if fwhm_nm_val else 'N/A'
    fnm = f'{fwhm_nm_val:.3f}' if fwhm_nm_val else 'N/A'
    fg = f'{fwhm_g:.1f}' if fwhm_g else 'N/A'
    u = f'{unc:.4f}' if not np.isnan(unc) else 'N/A'
    print(f"  {label_c:<8}  {'':>22}  {fg:>12}  {fnm:>12}  {L:>12}  {u:>14}")

print(f"""
  {'─'*65}
  📐 CAUSA FÍSICA — No es un problema del interrogador:

  La relación entre FWHM y longitud del FBG es:
      L ≈ λ_B² / (2·n_g·FWHM)   →  FWHM ∝ 1/L

  Si FWHM_POLY / FWHM_DW3 ≈ {(fwhm_g1/fwhm_g0):.2f}x, entonces:
      L_POLY ≈ L_DW3 / {(fwhm_g1/fwhm_g0):.2f} → FBG de POLY-1 es ~{(fwhm_g1/fwhm_g0):.1f}x más corto

  Posibles razones:
    1. FBG de POLY-1 grabado con menos periodos (proceso UV diferente)
    2. Fibra de poliamida con mayor chirp térmico/mecánico en el grating
    3. El canal 1 ve múltiples picos solapados (check espectro completo)

  {'─'*65}
  🎯 PRECISIÓN DEL CENTROIDE — Fit Gaussiano vs Interrogador:

  El interrogador calcula el centroide dentro de una ventana fija.
  Si el pico es más ancho, la ventana puede incluir ruido/lóbulos
  y el centroide es menos estable (mayor jitter en wav_peak).

  Solución: usar el μ del fit Gaussiano propio en lugar de wav_peak:
    • Canal 0 centroide fit: índice {mu0:.2f}  (±{unc0:.4f} pts)
    • Canal 1 centroide fit: índice {mu1:.2f}  (±{unc1:.4f} pts)

  Para obtener λ en nm necesitas calibrar: λ = λ_start + idx·Δλ
  (parámetros disponibles en la configuración del interrogador)
{'═'*65}""")
```
